## 1. Relational Database Connection


```python
import sqlite3
import pandas as pd
from pathlib import Path

db_path = Path('../database/silla.db')
conn = sqlite3.connect(db_path)
```

## 2. Product Brand Price Distribution


```python
query_brands = '''
SELECT brand, category, COUNT(*) as product_count, ROUND(AVG(price_egp), 2) as avg_price, MIN(price_egp) as min_price, MAX(price_egp) as max_price
FROM products
GROUP BY brand, category
ORDER BY product_count DESC
LIMIT 20;
'''
pd.read_sql(query_brands, conn)
```

## 3. Recipe-to-Cart Demonstration: Macaroni Béchamel (R0001)


```python
query_cart = '''
SELECT 
    r.recipe_name_en,
    ri.original_ingredient_name,
    ri.quantity as needed_qty,
    ri.unit as recipe_unit,
    p.product_name,
    p.brand,
    p.package_size,
    p.package_unit,
    p.price_egp,
    p.store_name
FROM recipes r
JOIN recipe_ingredients ri ON r.recipe_id = ri.recipe_id
JOIN ingredients i ON ri.ingredient_id = i.ingredient_id
JOIN products p ON i.ingredient_id = p.ingredient_id
WHERE r.recipe_id = 'R0001'
GROUP BY ri.original_ingredient_name
ORDER BY p.price_egp ASC;
'''
pd.read_sql(query_cart, conn)
```
