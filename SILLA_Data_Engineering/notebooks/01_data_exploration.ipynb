## 1. Environment Setup & Data Ingestion
Load processed datasets from the `data/processed` directory.


```python
import pandas as pd
import numpy as np
from pathlib import Path

data_dir = Path('../data/processed')
recipes = pd.read_csv(data_dir / 'recipes_clean.csv')
ingredients = pd.read_csv(data_dir / 'ingredients_master.csv')
recipe_ing = pd.read_csv(data_dir / 'recipe_ingredients_clean.csv')
products = pd.read_csv(data_dir / 'products_clean.csv')

print(f'Recipes: {recipes.shape}')
print(f'Ingredients Master: {ingredients.shape}')
print(f'Recipe Ingredients: {recipe_ing.shape}')
print(f'Products: {products.shape}')
```
